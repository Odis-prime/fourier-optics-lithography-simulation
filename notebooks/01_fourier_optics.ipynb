# Chapter 1: Mathematical Foundations of Fourier Optics

## 1.1 Electromagnetic Wave Propagation

Light is an electromagnetic wave whose propagation is governed by Maxwell's equations.

In a homogeneous, isotropic, source-free vacuum, Maxwell's equations take the form:

$$
\nabla \cdot \mathbf{E} = 0
$$

$$
\nabla \cdot \mathbf{B} = 0
$$

$$
\nabla \times \mathbf{E}
= -\frac{\partial \mathbf{B}}{\partial t}
$$

$$
\nabla \times \mathbf{B}
= \mu_0\epsilon_0\frac{\partial \mathbf{E}}{\partial t}
$$

Here, $\mathbf{E}$ and $\mathbf{B}$ denote the electric and magnetic fields, while $\epsilon_0$ and $\mu_0$ are the vacuum permittivity and permeability.

### 1.2 Derivation of the Electromagnetic Wave Equation

Taking the curl of Faraday's law:

$$
\nabla\times(\nabla\times\mathbf{E})
=-\frac{\partial}{\partial t}(\nabla\times\mathbf{B})
$$

Using the vector identity:

$$
\nabla\times(\nabla\times\mathbf{E})
=\nabla(\nabla\cdot\mathbf{E})-\nabla^2\mathbf{E}
$$

Since $\nabla\cdot\mathbf{E}=0$, we obtain:

$$
-\nabla^2\mathbf{E}
=-\mu_0\epsilon_0\frac{\partial^2\mathbf{E}}{\partial t^2}
$$

Therefore:

$$
\boxed{
\nabla^2\mathbf{E}
-\frac{1}{c^2}\frac{\partial^2\mathbf{E}}{\partial t^2}=0
}
$$

where:

$$
c=\frac{1}{\sqrt{\mu_0\epsilon_0}}
$$

This is the electromagnetic wave equation in vacuum.



### 1.3 Monochromatic Electromagnetic Waves

A monochromatic electromagnetic wave oscillates at a single angular frequency $\omega$. Its electric field can be represented using complex notation:

$$
\mathbf{E}(\mathbf{r},t)
=\operatorname{Re}\left\{
\mathbf{E}_0(\mathbf{r})e^{-i\omega t}
\right\}
$$

Here, $\mathbf{E}_0(\mathbf{r})$ is the complex spatial amplitude and $\omega=2\pi f$ is the angular frequency.

The electric field satisfies the wave equation in vacuum:

$$
\nabla^2\mathbf{E}
-\frac{1}{c^2}\frac{\partial^2\mathbf{E}}{\partial t^2}=0
$$

Using the complex representation, the second time derivative becomes:

$$
\frac{\partial^2}{\partial t^2}
\left(\mathbf{E}_0e^{-i\omega t}\right)
=-\omega^2\mathbf{E}_0e^{-i\omega t}
$$

Substituting into the wave equation and cancelling the common exponential factor yields:

$$
\nabla^2\mathbf{E}_0
+\frac{\omega^2}{c^2}\mathbf{E}_0=0
$$

Introducing the wavenumber:

$$
k=\frac{\omega}{c}=\frac{2\pi}{\lambda}
$$

we obtain the vector Helmholtz equation:

$$
\boxed{\nabla^2\mathbf{E}_0+k^2\mathbf{E}_0=0}
$$

For a single Cartesian component of the electric field, the scalar Helmholtz equation is:

$$
\boxed{\nabla^2U+k^2U=0}
$$

This equation describes the spatial distribution of a monochromatic optical field and provides the mathematical foundation for the diffraction and Fourier optics models developed in this project.



### 1.4 Plane Wave Solution of the Helmholtz Equation

To understand optical wave propagation, we begin with one of the simplest solutions of the Helmholtz equation: a monochromatic plane wave.

The scalar Helmholtz equation is given by:

$$
\nabla^2 U+k^2U=0
$$

#### Choice of the Propagation Direction

In optical systems, it is convenient to define the $z$-axis as the optical axis, corresponding to the main direction of light propagation.

This choice is not a physical restriction, since the coordinate system can be rotated to align with the propagation direction of a plane wave.

In optical lithography, the photomask is typically represented by a plane perpendicular to the optical axis. We therefore place the mask at $z=0$ and describe the transmitted field as it propagates toward positive $z$.

The transverse coordinates $x$ and $y$ describe the spatial distribution of the optical field across the mask.

#### Why Do We Consider a Plane Wave?

A plane wave provides a simple starting point for studying diffraction and optical imaging.

For a plane wave propagating along the positive $z$-direction, the wavevector is:

$$
\mathbf{k}=k\hat{\mathbf{z}}
$$

The spatial part of the electric field can then be written as:

$$
\boxed{U(z)=U_0e^{ikz}}
$$

where $U_0$ is the constant complex amplitude and $k=2\pi/\lambda$ is the wavenumber.

This solution is particularly useful because:

1. It represents a monochromatic wave with uniform amplitude across planes perpendicular to the propagation direction.
2. It describes an idealized form of coherent illumination incident normally on a photomask.
3. More complicated optical fields can be represented as superpositions of plane waves with different propagation directions, forming the basis of Fourier optics.

Thus, the plane-wave solution provides a natural starting point for the analysis of diffraction, Fourier transforms, and optical lithography.



### 1.5 Interaction of a Plane Wave with a Photomask

#### 1.5.1 Incident Optical Field

We consider a monochromatic plane wave propagating along the positive $z$-direction, normally incident on a photomask located at $z=0$.

Before reaching the photomask, the spatial part of the optical field is described by:

$$
U_{\mathrm{inc}}(x,y,z)=U_0e^{ikz}
$$

Since the amplitude is uniform across planes perpendicular to the propagation direction, the incident field has no transverse spatial dependence on $x$ or $y$.

Therefore:

$$
\frac{\partial U_{\mathrm{inc}}}{\partial x}=0,
\qquad
\frac{\partial U_{\mathrm{inc}}}{\partial y}=0
$$

The optical field depends only on the propagation coordinate $z$.

#### 1.5.2 Transmission Through the Photomask

A photomask contains a spatial pattern that modifies the amplitude and/or phase of the incident optical field.

Within the thin-mask approximation, its optical properties are described by a complex transmission function:

$$
t(x,y)=A(x,y)e^{i\phi(x,y)}
$$

where:

- $A(x,y)$ represents the amplitude transmission.
- $\phi(x,y)$ represents the phase shift introduced by the mask.

Immediately after the photomask, the transmitted field is:

$$
U(x,y,0^+)=t(x,y)U_{\mathrm{inc}}(x,y,0^-)
$$

Since the incident wave is normally incident and

$$
U_{\mathrm{inc}}(x,y,0^-)=U_0,
$$

we obtain:

$$
\boxed{U(x,y,0^+)=U_0t(x,y)}
$$

Consequently, the optical field immediately after the photomask generally acquires a transverse spatial dependence on $x$ and $y$.

This dependence is determined by the transmission pattern of the mask.

#### 1.5.3 Optical Field After the Photomask

Before the photomask, the incident optical field is a single plane wave:

$$
U_{\mathrm{inc}}(z)=U_0e^{ikz}
$$

After transmission through a nonuniform photomask, the field can no longer generally be described by a single plane wave propagating along the $z$-axis.

Instead, the transmitted field has the general spatial form:

$$
\boxed{U=U(x,y,z)}
$$

As the field propagates beyond the mask, different parts of the optical wave interfere, producing diffraction patterns.

The resulting optical distribution is governed by the Helmholtz equation:

$$
\nabla^2 U+k^2U=0
$$

or, in Cartesian coordinates:

$$
\boxed{
\frac{\partial^2U}{\partial x^2}
+
\frac{\partial^2U}{\partial y^2}
+
\frac{\partial^2U}{\partial z^2}
+k^2U=0
}
$$

The transverse structure introduced by the photomask can be analyzed by decomposing the transmitted field into a superposition of plane waves with different propagation directions.

This decomposition is mathematically described by the spatial Fourier transform and forms the foundation of Fourier optics.

#### 1.5.4 Relevance to Optical Lithography

In optical lithography, the photomask encodes the pattern intended to be transferred onto a photoresist-coated semiconductor wafer.

Diffraction and interference modify the spatial intensity distribution during propagation and imaging.

Consequently, the optical intensity reaching the photoresist is not necessarily an exact reproduction of the photomask pattern.

Understanding this transformation is essential for predicting lithographic image quality and resolution.



### 1.6 Mathematical Description of a Single-Slit Photomask

We consider an ideal, infinitely thin, opaque photomask located at the plane $z=0$.

The mask contains a single transparent slit of width $a$, centered at $x=0$ and extending infinitely along the $y$-direction.

Since the geometry is invariant along the $y$-axis, the problem can be treated as two-dimensional in the $(x,z)$ plane.

The amplitude transmission function of the photomask is defined as:

$$
\boxed{
M(x)=
\begin{cases}
1, & |x|\leq \frac{a}{2}\\
0, & |x|>\frac{a}{2}
\end{cases}
}
$$

The incident monochromatic plane wave propagates along the positive $z$-direction:

$$
U_{\mathrm{inc}}(z)=U_0e^{ikz}
$$

At the entrance plane of the photomask:

$$
U_{\mathrm{inc}}(x,0^-)=U_0
$$

Under the thin-mask approximation, the transmitted field immediately after the photomask is given by:

$$
U(x,0^+)=M(x)U_{\mathrm{inc}}(x,0^-)
$$

Therefore:

$$
\boxed{
U(x,0^+)=
\begin{cases}
U_0, & |x|\leq \frac{a}{2}\\
0, & |x|>\frac{a}{2}
\end{cases}
}
$$

This expression specifies the optical field at the exit plane of the mask.

For propagation into the region $z>0$, the field satisfies the scalar Helmholtz equation:

$$
\boxed{
\frac{\partial^2 U}{\partial x^2}
+
\frac{\partial^2 U}{\partial z^2}
+k^2U=0
}
$$

The transmitted field can be propagated by decomposing its transverse spatial distribution into plane-wave components using the Fourier transform.

The physically relevant solution is selected by considering waves propagating or decaying away from the mask.



### 1.7 Fourier Transform of the Photomask

The transmitted optical field immediately after the photomask is given by:

$$
U(x,0^+)=U_0M(x)
$$

To analyze its spatial frequency content, we introduce the one-dimensional Fourier transform:

$$
\widetilde{U}(k_x,0)=
\int_{-\infty}^{+\infty}
U(x,0^+)e^{-ik_xx}\,dx
$$

Here, $k_x$ denotes the transverse component of the wavevector.

For a single slit of width $a$, the transmission function is:

$$
M(x)=
\begin{cases}
1, & |x|\leq a/2\\
0, & |x|>a/2
\end{cases}
$$

Substituting the transmitted field into the Fourier transform:

$$
\widetilde{U}(k_x,0)
=U_0\int_{-a/2}^{a/2}e^{-ik_xx}\,dx
$$

Evaluating the integral gives:

$$
\widetilde{U}(k_x,0)
=
U_0\left[
\frac{e^{-ik_xx}}{-ik_x}
\right]_{-a/2}^{a/2}
$$

Using Euler's identity:

$$
e^{-i\alpha}-e^{i\alpha}=-2i\sin\alpha
$$

we obtain:

$$
\boxed{
\widetilde{U}(k_x,0)
=U_0a\,\operatorname{sinc}\left(\frac{k_xa}{2}\right)
}
$$

where the sinc function is defined as:

$$
\operatorname{sinc}(u)=\frac{\sin u}{u},
\qquad \operatorname{sinc}(0)=1
$$

#### Relationship Between Spatial Frequency and Diffraction Angle

For a plane-wave component propagating at an angle $\theta$ relative to the optical axis:

$$
k_x=k\sin\theta
$$

Since $k=2\pi/\lambda$, the Fourier spectrum becomes:

$$
\boxed{
\widetilde{U}(k\sin\theta,0)
=U_0a\,\operatorname{sinc}
\left(\frac{\pi a\sin\theta}{\lambda}\right)
}
$$

In the Fraunhofer approximation, the far-field amplitude is proportional to the Fourier transform of the transmitted field.

Therefore, the normalized angular intensity distribution is:

$$
\boxed{
\frac{I(\theta)}{I(0)}
=
\operatorname{sinc}^2
\left(\frac{\pi a\sin\theta}{\lambda}\right)
}
$$

This result demonstrates that the diffraction pattern produced by a single slit is determined by the Fourier transform of its transmission function.

The next step is to study the propagation of these spatial-frequency components beyond the photomask using the angular spectrum method.



### 1.8 Angular Spectrum Method

#### 1.8.1 Fourier Transformation of the Helmholtz Equation

The transmitted optical field propagates in the homogeneous region $z>0$, where it satisfies the scalar Helmholtz equation:

$$
\frac{\partial^2U}{\partial x^2}
+\frac{\partial^2U}{\partial z^2}
+k^2U=0
$$

Since the photomask is invariant along the $y$-direction, the problem is described in two spatial dimensions.

We define the Fourier transform of the optical field with respect to the transverse coordinate $x$:

$$
\widetilde U(k_x,z)
=
\int_{-\infty}^{+\infty}
U(x,z)e^{-ik_xx}\,dx
$$

Using the Fourier differentiation property:

$$
\mathcal F_x\left\{
\frac{\partial^2U}{\partial x^2}
\right\}
=-k_x^2\widetilde U
$$

and noting that the transform is performed only with respect to $x$, we obtain:

$$
-k_x^2\widetilde U
+\frac{\partial^2\widetilde U}{\partial z^2}
+k^2\widetilde U=0
$$

Therefore:

$$
\boxed{
\frac{\partial^2\widetilde U}{\partial z^2}
+(k^2-k_x^2)\widetilde U=0
}
$$

The partial differential equation has been reduced to an ordinary differential equation in $z$ for each transverse spatial frequency $k_x$.

#### 1.8.2 Solution in the Spatial-Frequency Domain

Defining the longitudinal wavevector component by:

$$
k_z^2=k^2-k_x^2
$$

the equation becomes:

$$
\frac{d^2\widetilde U}{dz^2}
+k_z^2\widetilde U=0
$$

Its general solution is:

$$
\widetilde U(k_x,z)
=A(k_x)e^{ik_zz}+B(k_x)e^{-ik_zz}
$$

We assume that no waves are incident from the region $z>0$. Therefore, only outgoing or exponentially decaying components are retained:

$$
\widetilde U(k_x,z)=A(k_x)e^{ik_zz}
$$

The longitudinal wavevector component is selected as:

$$
k_z=
\begin{cases}
\sqrt{k^2-k_x^2}, & |k_x|\leq k\\
i\sqrt{k_x^2-k^2}, & |k_x|>k
\end{cases}
$$

The first case describes propagating plane waves, while the second describes evanescent components that decay exponentially with distance.

At the photomask exit plane $z=0^+$:

$$
A(k_x)=\widetilde U(k_x,0)
$$

Consequently:

$$
\boxed{
\widetilde U(k_x,z)
=\widetilde U(k_x,0)e^{ik_zz}
}
$$

#### 1.8.3 Reconstruction of the Optical Field

Applying the inverse Fourier transform:

$$
U(x,z)=\frac{1}{2\pi}
\int_{-\infty}^{+\infty}
\widetilde U(k_x,z)e^{ik_xx}\,dk_x
$$

and substituting the propagation factor gives:

$$
\boxed{
U(x,z)=\frac{1}{2\pi}
\int_{-\infty}^{+\infty}
\widetilde U(k_x,0)
e^{i(k_xx+k_zz)}\,dk_x
}
$$

This expression is known as the Angular Spectrum Method.

It describes the propagation of a monochromatic optical field as a superposition of plane-wave components.

Each spatial-frequency component evolves independently according to its longitudinal wavevector $k_z$.

The method provides an exact propagation solution within the homogeneous scalar Helmholtz model, without requiring the paraxial approximation.

#### 1.8.4 Application to the Single-Slit Photomask

For the single-slit photomask introduced previously:

$$
U(x,0^+)=U_0M(x)
$$

Its Fourier transform is:

$$
\widetilde U(k_x,0)=
U_0a\,\operatorname{sinc}\left(\frac{k_xa}{2}\right)
$$

Substituting this spectrum into the Angular Spectrum integral yields:

$$
\boxed{
U(x,z)=\frac{U_0a}{2\pi}
\int_{-\infty}^{+\infty}
\operatorname{sinc}\left(\frac{k_xa}{2}\right)
e^{i(k_xx+k_zz)}\,dk_x
}
$$

where:

$$
k_z=\sqrt{k^2-k_x^2}
$$

with the outgoing-wave and evanescent-decay branch convention defined above.

The optical intensity is proportional to the squared magnitude of the complex field:

$$
\boxed{I(x,z)\propto |U(x,z)|^2}
$$

This formulation enables the calculation of diffraction patterns at arbitrary propagation distances and provides the theoretical basis for numerical wave propagation using the Fast Fourier Transform (FFT).
